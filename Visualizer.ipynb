```python
import pandas as pd
import matplotlib.pyplot as plt
import os


# =========================================================
# GLOBAL VARIABLE
# =========================================================

df = None
last_plot = None


# =========================================================
# MAIN MENU
# =========================================================

def main_menu():

    global df

    while True:

        print("\n")
        print("========== Data Analysis & Visualization Program ==========")
        print("Please select an option:")
        print("1. Load Dataset")
        print("2. Explore Data")
        print("3. Perform DataFrame Operations")
        print("4. Handle Missing Data")
        print("5. Generate Descriptive Statistics")
        print("6. Data Visualization")
        print("7. Save Visualization")
        print("8. Exit")
        print("==========================================================")

        choice = input("Enter your choice: ")

        if choice == "1":
            load_dataset()

        elif choice == "2":
            explore_data()

        elif choice == "3":
            dataframe_operations()

        elif choice == "4":
            missing_data()

        elif choice == "5":
            descriptive_statistics()

        elif choice == "6":
            data_visualization()

        elif choice == "7":
            save_visualization()

        elif choice == "8":
            print("\nExiting the program. Goodbye!")
            break

        else:
            print("\nInvalid choice! Please enter a number from 1 to 8.")


# =========================================================
# 1. LOAD DATASET
# =========================================================

def load_dataset():

    global df

    print("\n== Load Dataset ==")

    path = input("Enter the path of the dataset (CSV file): ")

    if not os.path.exists(path):
        print("File not found!")
        return

    try:
        df = pd.read_csv(path)

        print("Dataset loaded successfully!")

        print("\nDataset Preview:")
        print(df.head())

    except Exception as e:
        print("Error loading dataset:", e)


# =========================================================
# CHECK DATASET
# =========================================================

def check_dataset():

    if df is None:
        print("\nPlease load the dataset first using option 1.")
        return False

    return True


# =========================================================
# 2. EXPLORE DATA
# =========================================================

def explore_data():

    if not check_dataset():
        return

    while True:

        print("\n== Explore Data ==")
        print("1. Display the first 5 rows")
        print("2. Display the last 5 rows")
        print("3. Display column names")
        print("4. Display data types")
        print("5. Display basic info")
        print("6. Back to Main Menu")

        choice = input("Enter your choice: ")

        if choice == "1":

            print("\nFirst 5 Rows:")
            print(df.head())

        elif choice == "2":

            print("\nLast 5 Rows:")
            print(df.tail())

        elif choice == "3":

            print("\nColumn Names:")
            print(list(df.columns))

        elif choice == "4":

            print("\nData Types:")
            print(df.dtypes)

        elif choice == "5":

            print("\nBasic Information:")
            df.info()

        elif choice == "6":
            break

        else:
            print("Invalid choice!")


# =========================================================
# 3. DATAFRAME OPERATIONS
# =========================================================

def dataframe_operations():

    global df

    if not check_dataset():
        return

    while True:

        print("\n== Perform DataFrame Operations ==")
        print("1. Select a column")
        print("2. Add a new column")
        print("3. Delete a column")
        print("4. Sort DataFrame")
        print("5. Filter Data")
        print("6. Display DataFrame shape")
        print("7. Display DataFrame")
        print("8. Back to Main Menu")

        choice = input("Enter your choice: ")

        # -----------------------------------------
        # SELECT COLUMN
        # -----------------------------------------

        if choice == "1":

            column = input("Enter column name: ")

            if column in df.columns:
                print("\nSelected Column:")
                print(df[column])
            else:
                print("Column not found!")

        # -----------------------------------------
        # ADD COLUMN
        # -----------------------------------------

        elif choice == "2":

            column = input("Enter new column name: ")

            value = input("Enter value for the new column: ")

            try:
                df[column] = value
                print("Column added successfully!")
                print(df.head())

            except Exception as e:
                print("Error:", e)

        # -----------------------------------------
        # DELETE COLUMN
        # -----------------------------------------

        elif choice == "3":

            column = input("Enter column name to delete: ")

            if column in df.columns:

                df.drop(column, axis=1, inplace=True)

                print("Column deleted successfully!")

            else:
                print("Column not found!")

        # -----------------------------------------
        # SORT
        # -----------------------------------------

        elif choice == "4":

            column = input("Enter column name for sorting: ")

            if column in df.columns:

                order = input("Enter order (ascending/descending): ")

                if order.lower() == "descending":
                    df = df.sort_values(by=column, ascending=False)
                else:
                    df = df.sort_values(by=column, ascending=True)

                print("\nSorted DataFrame:")
                print(df)

            else:
                print("Column not found!")

        # -----------------------------------------
        # FILTER
        # -----------------------------------------

        elif choice == "5":

            column = input("Enter column name for filtering: ")

            if column not in df.columns:
                print("Column not found!")
                continue

            value = input("Enter value to filter: ")

            filtered = df[df[column].astype(str).str.lower() ==
                          value.lower()]

            print("\nFiltered Data:")
            print(filtered)

        # -----------------------------------------
        # SHAPE
        # -----------------------------------------

        elif choice == "6":

            print("\nRows:", df.shape[0])
            print("Columns:", df.shape[1])

        # -----------------------------------------
        # DISPLAY
        # -----------------------------------------

        elif choice == "7":

            print("\nDataFrame:")
            print(df)

        elif choice == "8":
            break

        else:
            print("Invalid choice!")


# =========================================================
# 4. HANDLE MISSING DATA
# =========================================================

def missing_data():

    global df

    if not check_dataset():
        return

    while True:

        print("\n== Handle Missing Data ==")
        print("1. Display rows with missing values")
        print("2. Fill missing values with mean")
        print("3. Drop rows with missing values")
        print("4. Replace missing values with a specific value")
        print("5. Back to Main Menu")

        choice = input("Enter your choice: ")

        # -----------------------------------------
        # DISPLAY MISSING
        # -----------------------------------------

        if choice == "1":

            missing_rows = df[df.isnull().any(axis=1)]

            if missing_rows.empty:
                print("\nNo missing values found in the dataset!")
            else:
                print("\nRows with missing values:")
                print(missing_rows)

        # -----------------------------------------
        # FILL WITH MEAN
        # -----------------------------------------

        elif choice == "2":

            numeric_columns = df.select_dtypes(
                include="number"
            ).columns

            if len(numeric_columns) == 0:

                print("No numeric columns found!")

            else:

                for column in numeric_columns:
                    df[column] = df[column].fillna(
                        df[column].mean()
                    )

                print("Missing numeric values filled with mean!")

        # -----------------------------------------
        # DROP MISSING
        # -----------------------------------------

        elif choice == "3":

            before = len(df)

            df.dropna(inplace=True)

            after = len(df)

            print("Rows removed:", before - after)

        # -----------------------------------------
        # SPECIFIC VALUE
        # -----------------------------------------

        elif choice == "4":

            value = input(
                "Enter the value to replace missing values: "
            )

            df.fillna(value, inplace=True)

            print("Missing values replaced successfully!")

        elif choice == "5":
            break

        else:
            print("Invalid choice!")


# =========================================================
# 5. DESCRIPTIVE STATISTICS
# =========================================================

def descriptive_statistics():

    if not check_dataset():
        return

    print("\n== Descriptive Statistics ==")

    print("\n1. Count")
    print("2. Mean")
    print("3. Median")
    print("4. Minimum")
    print("5. Maximum")
    print("6. Standard Deviation")
    print("7. Complete Statistics")
    print("8. Back to Main Menu")

    choice = input("Enter your choice: ")

    numeric_data = df.select_dtypes(include="number")

    if numeric_data.empty:

        print("No numeric columns found!")
        return

    if choice == "1":

        print("\nCount:")
        print(numeric_data.count())

    elif choice == "2":

        print("\nMean:")
        print(numeric_data.mean())

    elif choice == "3":

        print("\nMedian:")
        print(numeric_data.median())

    elif choice == "4":

        print("\nMinimum:")
        print(numeric_data.min())

    elif choice == "5":

        print("\nMaximum:")
        print(numeric_data.max())

    elif choice == "6":

        print("\nStandard Deviation:")
        print(numeric_data.std())

    elif choice == "7":

        print("\nComplete Descriptive Statistics:")
        print(numeric_data.describe())

    elif choice == "8":
        return

    else:
        print("Invalid choice!")


# =========================================================
# 6. DATA VISUALIZATION
# =========================================================

def data_visualization():

    global last_plot

    if not check_dataset():
        return

    print("\n== Data Visualization ==")
    print("1. Bar plot")
    print("2. Line plot")
    print("3. Scatter plot")
    print("4. Pie chart")
    print("5. Histogram")
    print("6. Stack plot")

    choice = input("Enter your choice: ")

    # =====================================================
    # BAR PLOT
    # =====================================================

    if choice == "1":

        print("\n== Bar Plot ==")

        x_column = input("Enter x-axis column name: ")
        y_column = input("Enter y-axis column name: ")

        if x_column not in df.columns or y_column not in df.columns:
            print("Invalid column name!")
            return

        plt.figure(figsize=(8, 5))

        plt.bar(
            df[x_column].astype(str),
            df[y_column]
        )

        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.title("Bar Plot")

        plt.xticks(rotation=45)
        plt.tight_layout()

        print("Generating bar plot...")

        plt.show()

        last_plot = plt.gcf()

        print("Bar plot displayed successfully!")

    # =====================================================
    # LINE PLOT
    # =====================================================

    elif choice == "2":

        print("\n== Line Plot ==")

        x_column = input("Enter x-axis column name: ")
        y_column = input("Enter y-axis column name: ")

        if x_column not in df.columns or y_column not in df.columns:
            print("Invalid column name!")
            return

        plt.figure(figsize=(8, 5))

        plt.plot(
            df[x_column],
            df[y_column],
            marker="o"
        )

        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.title("Line Plot")

        plt.xticks(rotation=45)
        plt.tight_layout()

        print("Generating line plot...")

        plt.show()

        last_plot = plt.gcf()

        print("Line plot displayed successfully!")

    # =====================================================
    # SCATTER PLOT
    # =====================================================

    elif choice == "3":

        print("\n== Scatter Plot ==")

        x_column = input("Enter x-axis column name: ")
        y_column = input("Enter y-axis column name: ")

        if x_column not in df.columns or y_column not in df.columns:
            print("Invalid column name!")
            return

        plt.figure(figsize=(8, 5))

        plt.scatter(
            df[x_column],
            df[y_column]
        )

        plt.xlabel(x_column)
        plt.ylabel(y_column)
        plt.title("Scatter Plot")

        plt.tight_layout()

        print("Generating scatter plot...")

        plt.show()

        last_plot = plt.gcf()

        print("Scatter plot displayed successfully!")

    # =====================================================
    # PIE CHART
    # =====================================================

    elif choice == "4":

        print("\n== Pie Chart ==")

        column = input(
            "Enter column name for pie chart: "
        )

        if column not in df.columns:
            print("Invalid column name!")
            return

        counts = df[column].value_counts()

        plt.figure(figsize=(7, 7))

        plt.pie(
            counts.values,
            labels=counts.index,
            autopct="%1.1f%%"
        )

        plt.title("Pie Chart")

        print("Generating pie chart...")

        plt.show()

        last_plot = plt.gcf()

        print("Pie chart displayed successfully!")

    # =====================================================
    # HISTOGRAM
    # =====================================================

    elif choice == "5":

        print("\n== Histogram ==")

        column = input(
            "Enter column name for histogram: "
        )

        if column not in df.columns:
            print("Invalid column name!")
            return

        if not pd.api.types.is_numeric_dtype(df[column]):
            print("Histogram requires a numeric column!")
            return

        plt.figure(figsize=(8, 5))

        plt.hist(
            df[column].dropna(),
            bins=10
        )

        plt.xlabel(column)
        plt.ylabel("Frequency")
        plt.title("Histogram")

        plt.tight_layout()

        print("Generating histogram...")

        plt.show()

        last_plot = plt.gcf()

        print("Histogram displayed successfully!")

    # =====================================================
    # STACK PLOT
    # =====================================================

    elif choice == "6":

        print("\n== Stack Plot ==")

        column1 = input(
            "Enter first numeric column name: "
        )

        column2 = input(
            "Enter second numeric column name: "
        )

        if column1 not in df.columns or column2 not in df.columns:
            print("Invalid column name!")
            return

        if not pd.api.types.is_numeric_dtype(
            df[column1]
        ) or not pd.api.types.is_numeric_dtype(
            df[column2]
        ):

            print("Stack plot requires numeric columns!")
            return

        x = range(len(df))

        plt.figure(figsize=(8, 5))

        plt.stackplot(
            x,
            df[column1].fillna(0),
            df[column2].fillna(0),
            labels=[column1, column2]
        )

        plt.xlabel("Index")
        plt.ylabel("Values")
        plt.title("Stack Plot")
        plt.legend()

        plt.tight_layout()

        print("Generating stack plot...")

        plt.show()

        last_plot = plt.gcf()

        print("Stack plot displayed successfully!")

    else:

        print("Invalid visualization choice!")


# =========================================================
# 7. SAVE VISUALIZATION
# =========================================================

def save_visualization():

    global last_plot

    print("\n== Save Visualization ==")

    if last_plot is None:

        print("No visualization available to save!")
        print("Please generate a visualization first using option 6.")
        return

    filename = input(
        "Enter file name to save the plot "
        "(e.g. scatter_plot.png): "
    )

    if filename.strip() == "":
        print("Invalid file name!")
        return

    try:

        last_plot.savefig(
            filename,
            dpi=300,
            bbox_inches="tight"
        )

        print(
            "Visualization saved as",
            filename,
            "successfully!"
        )

    except Exception as e:

        print("Error saving visualization:", e)


# =========================================================
# START PROGRAM
# =========================================================

if __name__ == "__main__":
    main_menu()
```
